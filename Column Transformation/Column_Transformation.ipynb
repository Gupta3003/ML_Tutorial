```python
import pandas as pd
data = pd.read_csv('covid_people_database.csv', encoding='latin1')
data.head()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
for col in data.columns:
  print(data[col].value_counts())
```

### Age - Numerical Data
### Gender - Nomial Data
### Fever - Numerical Data with Na value
### Cough - Nomial Data
### City - Nomial Data
### Fever Level - Ordinal Data

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.iloc[:,0:6], data.iloc[:,-1], test_size=0.2, random_state=2)
```

# OneHotEncoding - Gender Cough City
# Ordinal Encoding - Fever Level
# Imputer - Fever

# Long Method

## Imputer

```python
si=SimpleImputer()
X_train_fever=si.fit_transform(X_train[['fever']])
X_test_fever=si.fit_transform(X_test[['fever']])
X_train_fever.shape
```

## Ordinal Encoding

```python
oe=OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
X_train_fever_level = oe.fit_transform(X_train[['fever_level']])
X_test_fever_level = oe.fit_transform(X_test[['fever_level']])
X_train_fever_level.shape
```

## OneHotEncoding

```python
ohe=OneHotEncoder(drop='first',sparse_output=False)
# Gender, Age and City transformation
X_train_encode3=ohe.fit_transform(X_train[['gender', 'cough', 'city']])
X_test_encode3=ohe.fit_transform(X_test[['gender', 'cough', 'city']])
X_train_encode3.shape
```

## Extracting the Age Column

```python
X_train_age=X_train.drop(columns=['gender', 'fever', 'cough', 'city', 'fever_level']).values
X_test_age=X_test.drop(columns=['gender', 'fever', 'cough', 'city', 'fever_level']).values
X_train_age.shape
```

## Merge all column of X_train tranformation

```python
import numpy as np
np.concatenate((X_train_age, X_train_encode3, X_train_fever, X_train_fever_level), axis=1)
np.concatenate((X_test_age, X_test_encode3, X_test_fever, X_test_fever_level), axis=1)
```

```python
np.concatenate((X_train_age, X_train_encode3, X_train_fever, X_train_fever_level), axis=1).shape
```

# Short and Easy Method

```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(),['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Low', 'Medium', 'High']]), ['fever_level']),
    ('tnf3', OneHotEncoder(drop='first', sparse_output=False), ['gender', 'cough', 'city'])
], remainder='passthrough')
```

```python
transformer.fit_transform(X_train)
```
