```python
data=data.iloc[:,2:]
data.head()
```

```python
for col in data.columns :
  print(data[col].value_counts())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.iloc[:,0:2], data.iloc[:,-1], test_size=0.2, random_state=2)
```

```python
X_train.columns
```

```python
y_train
```

```python
from sklearn.preprocessing import OrdinalEncoder
os=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
os.fit(X_train)
```

```python
X_train_encode=os.transform(X_train)
X_test_encode=os.transform(X_test)
```

```python
os.categories_
```

```python
X_train=pd.DataFrame(data=X_train_encode , columns=X_train.columns)
X_test=pd.DataFrame(data=X_test_encode , columns=X_test.columns)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le.fit(y_train)
```

```python
y_train_encode=le.transform(y_train)
y_test_encode=le.transform(y_test)
```
