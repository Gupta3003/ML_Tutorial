# One Hot Encoding using sklearn in python

```python
data.head()
```

```python
X = data[['brand', 'year', 'km_driven', 'fuel', 'seller_type',
          'transmission', 'owner']]
y = data['selling_price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

### ohe=OneHotEncoder(drop='first'/'ifbinary', sparse_output=True/False, dtype=np.int64)
### when sparse_output=True then it give me the result in 1D Array
### when spase_output=False then it give me the result in 2D Array
### drop='first' remove the first column in the encoding columns
### drop='ifbinary' remove if feature has two value only
### we can dtype to chnage the data types


```python
# ohe=OneHotEncoder()
# ohe=OneHotEncoder(drop='first', sparse=True/False)
ohe=OneHotEncoder(drop='first',sparse_output=False)
ohe.fit_transform(X_train[['fuel', 'transmission', 'seller_type']])
ohe.fit_transform(X_test[['fuel', 'transmission', 'seller_type']])
```

```python
ohe.fit_transform(X_train[['fuel', 'transmission', 'seller_type']])
ohe.fit_transform(X_test[['fuel', 'transmission', 'seller_type']])
```

### if sparse_output=True then use convert it into 2d array by using toarray() function in this code
### ohe.fit_transform(X_train[['fuel', 'transmission', 'seller_type']]).toarray()

```python
X_train_new=ohe.fit_transform(X_train[['fuel', 'transmission', 'seller_type']])
X_test_new=ohe.fit_transform(X_test[['fuel', 'transmission', 'seller_type']])
```

```python
import numpy as np
X_train_final = np.hstack((
    X_train[['brand', 'year', 'km_driven', 'owner']].values,
    X_train_new
))

X_test_final = np.hstack((
    X_test[['brand', 'year', 'km_driven', 'owner']].values,
    X_test_new
))
```
